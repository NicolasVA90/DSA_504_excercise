## 1. Variables and Data Types

A **variable** is a name that points to a value stored in memory. In Python:
- You don't have to declare a type in advance — Python figures it out automatically. This is called **dynamic typing**.
- Variable names can contain letters, numbers, and underscores, but can't start with a number, and are case-sensitive (`revenue` and `Revenue` are different variables).
- By convention, Python variable names use `snake_case` (words separated by underscores), not `camelCase`.

Let's start with the four types you'll use constantly this semester.


```python
# Creating variables of different types
store_name = "Utica Main ..."        # str  -- text, always in quotes
daily_revenue = 4521.75             # float -- a number with a decimal point
units_sold = 138                    # int   -- a whole number, no decimal
is_open_today = True                # bool  -- only ever True or False

print(store_name)
print(daily_revenue)
print(units_sold)
print(is_open_today)
```

### Checking a variable's type

The built-in `type()` function tells you exactly what kind of value you're working with. This becomes essential once we're inspecting real spreadsheet columns in a few weeks — a column that *looks* numeric might secretly be stored as text, and `type()` (or pandas' equivalent, `.dtype`) is how you catch that.


```python
# Creating variables of different types
store_name = "Utica Main St..."        # str  -- text, always in quotes
daily_revenue = 4521.75             # float -- a number with a decimal point
units_sold = 138                    # int   -- a whole number, no decimal
is_open_today = True                # bool  -- only ever True or False

print(type(store_name))     # <class 'str'>
print(type(daily_revenue))  # <class 'float'>
print(type(units_sold))     # <class 'int'>
print(type(is_open_today))  # <class 'bool'>
```

### A fifth type worth knowing now: `None`

`None` represents "no value at all" — not zero, not an empty string, but a genuine absence of data. This is exactly what a missing cell in a spreadsheet becomes once we load it into Python in a few weeks (pandas calls it `NaN`, which behaves similarly).


```python
last_inspection_date = None
print(last_inspection_date)       # None
print(type(last_inspection_date)) # <class 'NoneType'>

# None is not the same as an empty string or zero
print(None == "")   # False
print(None == 0)    # False
```

### Type conversion (casting)

Sometimes a value arrives as the wrong type for what you need — very common with real-world data, where numbers are often read in as text. Python lets you convert between types explicitly using `int()`, `float()`, `str()`, and `bool()`.


```python
# Converting between types
units_as_text = "138"          # this is a string, not a number
print(type(units_as_text))     # <class 'str'>

units_as_number = int(units_as_text)   # now it's an actual integer
print(type(units_as_number))           # <class 'int'>
print(units_as_number + 10)            # 148 -- math works now

# float <-> int conversion
price = 19.99
price_rounded = int(price)      # int() truncates, it does NOT round
print(price_rounded)            # 19

# Numbers to strings, for building messages
revenue = 4521.75
message = "Revenue: $" + str(revenue)   # str() lets you concatenate a number into text
print(message)
```

```python
# A common real-world trap: int() fails on text that isn't a clean number
messy_value = "138 units"

try:
    fixed = int(messy_value)
except ValueError as e:
    print(f"Conversion failed: {e}")

# This is exactly the kind of problem we'll spend Classes 6-7 learning to clean up
```

### Dynamic typing in action

Because Python doesn't lock a variable to one type, the *same variable name* can point to a different type later in your program. This is flexible, but it's also a common source of bugs if you're not paying attention.


```python
value = 138          # value is currently an int
print(type(value))   # <class 'int'>

value = "138 units"   # now value points to a string instead
print(type(value))    # <class 'str'>

value = 138.0          # now a float
print(type(value))     # <class 'float'>
```

## 2. Arithmetic Operators

Python supports the full set of arithmetic operators, not just `+` and `-`:

| Operator | Meaning | Example |
|---|---|---|
| `+` | Addition | `5 + 2` → `7` |
| `-` | Subtraction | `5 - 2` → `3` |
| `*` | Multiplication | `5 * 2` → `10` |
| `/` | Division (always returns a float) | `5 / 2` → `2.5` |
| `//` | Floor division (rounds down to whole number) | `5 // 2` → `2` |
| `%` | Modulo (the remainder after division) | `5 % 2` → `1` |
| `**` | Exponent (power) | `5 ** 2` → `25` |


**Where you'll actually use `%` and `//`:** modulo is the standard trick for "every Nth item" logic — for example, flagging every 5th row in a dataset, or checking if a number is even (`n % 2 == 0`). Floor division is useful for things like "how many full weeks are in this many days."


```python
# Every arithmetic operator, demonstrated
a = 17
b = 5

print(a + b)    # 22   addition
print(a - b)    # 12   subtraction
print(a * b)    # 85   multiplication
print(a / b)    # 3.4  division -- always a float, even if it divides evenly
print(a // b)   # 3    floor division -- rounds DOWN to the nearest whole number
print(a % b)    # 2    modulo -- the remainder
print(a ** b)   # 1419857  exponent (17 to the power of 5)
```

```python
# Practical use of modulo: is a number even or odd?
for n in [10, 11, 12, 13]:
    if n % 2 == 0:
        print(f"{n} is even")
    else:
        print(f"{n} is odd")

# Practical use of floor division: convert days into weeks + leftover days
total_days = 17
weeks = total_days // 7
leftover_days = total_days % 7
print(f"{total_days} days = {weeks} full weeks and {leftover_days} leftover days")
```

### Order of operations

Python follows standard math order of operations (PEMDAS: Parentheses, Exponents, Multiplication/Division, Addition/Subtraction). Use parentheses whenever you want to be explicit — it costs nothing and prevents bugs.


```python
# Order of operations matters
result1 = 2 + 3 * 4        # multiplication happens first: 2 + 12 = 14
result2 = (2 + 3) * 4      # parentheses force addition first: 5 * 4 = 20

print(result1)  # 14
print(result2)  # 20

# A realistic example: computing average revenue per unit
total_revenue = 4521.75
total_units = 138
average_price_per_unit = total_revenue / total_units
print(round(average_price_per_unit, 2))
```

### String operators and f-strings

Strings support `+` (concatenation) and `*` (repetition) — but note these do something completely different from what they do with numbers.


```python
# String concatenation and repetition
store = "Utica Main St"
divider = "-" * 30    # repeats the string 20 times

print(store + " Sales Report")
print(divider)

# f-strings: the modern, readable way to build strings with variables inside
revenue = 4521.75
units = 138
print(f"{store} sold {units} units for ${revenue:.2f} today")

# f-strings also let you do quick formatting: :.2f means "2 decimal places"
pi_ish = 3.14159265
print(f"Pi rounded: {pi_ish:.2f}")
print(f"Pi with no rounding: {pi_ish}")
```

```python
# More f-string tricks you'll find useful
units = 138
price = 32.75

# Math directly inside an f-string
print(f"Total revenue: ${units * price:.2f}")

# Padding numbers for aligned output (useful for printed reports)
for u in [5, 42, 138]:
    print(f"Units: {u:>5}")   # right-align within 5 characters
```

## 3. Comparison and Logical Operators

**Comparison operators** always evaluate to a boolean (`True` or `False`):

| Operator | Meaning |
|---|---|
| `==` | equal to |
| `!=` | not equal to |
| `>` | greater than |
| `<` | less than |
| `>=` | greater than or equal to |
| `<=` | less than or equal to |

Note the difference between `=` (assignment — "set this variable") and `==` (comparison — "are these equal?"). Mixing these up is one of the most common beginner mistakes.


```python
units_sold = 138

print(units_sold == 138)   # True
print(units_sold != 138)   # False
print(units_sold > 100)    # True
print(units_sold < 100)    # False
print(units_sold >= 138)   # True
print(units_sold <= 137)   # False

# Comparisons work on strings too (alphabetical order)
print("apple" < "banana")  # True
print("Utica" == "utica")  # False -- comparisons are case-sensitive
```

### Logical operators: `and`, `or`, `not`

Logical operators combine multiple `True`/`False` values into a single result.

- `and` — True only if **both** sides are True
- `or` — True if **at least one** side is True
- `not` — flips True to False, and False to True


```python
units_sold = 138
is_weekend = True

# and -- both conditions must be true
print(units_sold > 100 and is_weekend)       # True and True -> True
print(units_sold > 200 and is_weekend)       # False and True -> False

# or -- at least one condition must be true
print(units_sold > 200 or is_weekend)        # False or True -> True

# not -- flips the boolean
print(not is_weekend)                        # False
```

```python
# A realistic combined example: flag a day as "notable" if it's either
# a very high units day, OR a weekend with above-average units
units_sold = 160
is_weekend = True
average_units = 120

notable_day = (units_sold > 200) or (is_weekend and units_sold > average_units)
print(notable_day)   # True, because the weekend + above-average condition is met
```

**Why this matters for data work:** filtering real datasets is almost entirely built from chained comparisons and logical operators — "give me all rows where revenue > 1000 AND category == 'electronics'" is the exact same logic you just wrote above, just applied to a table instead of a single variable. ## 4. Conditionals

Conditionals let your program make decisions and run different code depending on the situation.


```python
# Simple if -- runs only when the condition is True
units_sold = 138

if units_sold > 100:
    print("That's a solid number of units.")
```

```python
# if / else -- one of two paths always runs
units_sold = 45

if units_sold > 100:
    print("Strong sales day")
else:
    print("Below 100 units today")
```

```python
# if / elif / else -- multiple branches, checked in order
units_sold = 138

if units_sold > 150:
    print("Strong sales day")
elif units_sold > 75:
    print("Average sales day")
else:
    print("Slow sales day")
# Only ONE branch runs -- Python stops checking once it finds a True condition
```

### Nested conditionals

You can put an `if` statement inside another `if` statement, when a decision depends on more than one layer of logic.


```python
units_sold = 160
store_is_open = True

if store_is_open:
    if units_sold > 150:
        print("Great day — store was open and sales were strong")
    else:
        print("Store was open, but sales were modest")
else:
    print("Store was closed — no sales expected")
```

**A note on style:** the nested version above can often be rewritten using `and`, which is frequently more readable:


```python
# Same logic as above, written with 'and' instead of nesting
units_sold = 160
store_is_open = True

if store_is_open and units_sold > 150:
    print("Great day — store was open and sales were strong")
elif store_is_open:
    print("Store was open, but sales were modest")
else:
    print("Store was closed — no sales expected")
```

## 5. Loops

Loops let you repeat an action across many values — essential once we're working with hundreds or thousands of data rows instead of one variable at a time.


### `for` loops — iterating over a known sequence


```python
# for loop over a list
daily_units = [138, 95, 210, 60, 175]

for units in daily_units:
    print(units)
```

```python
# for loop with a condition inside
daily_units = [138, 95, 210, 60, 175]

for units in daily_units:
    if units > 150:
        print(f"{units} units — strong day")
    else:
        print(f"{units} units — normal day")
```

```python
# for loop over range() -- useful when you want to repeat N times,
# or need a counting number rather than looping over existing data
for day_number in range(1, 6):    # 1, 2, 3, 4, 5 (stops before 6)
    print(f"Day {day_number}")
```

```python
# for loop over a string -- strings are sequences too
store_code = "UTC1"
for character in store_code:
    print(character)
```

### `while` loops — repeating until a condition becomes False

Use `while` when you don't know in advance how many times you'll need to repeat — you're repeating *until something becomes true*, rather than a fixed number of times.


```python
# while loop -- runs until a condition is False
day_count = 0
running_total = 0
daily_units = [138, 95, 210, 60, 175]

while running_total < 300:
    running_total += daily_units[day_count]
    day_count += 1

print(f"Took {day_count} days to exceed 300 units (total: {running_total})")
```

### `break` and `continue`

- `break` exits the loop immediately, even if there was more to go through.
- `continue` skips the rest of the current iteration and jumps to the next one.


```python
# break -- stop the loop early once we find what we're looking for
daily_units = [95, 138, 210, 60, 175]

for units in daily_units:
    if units > 200:
        print(f"Found a day over 200 units: {units}. Stopping search.")
        break
    print(f"Checked {units}, not over 200 yet")
```

```python
# continue -- skip low-value days, only print the ones worth reporting
daily_units = [45, 138, 30, 210, 60, 175]

for units in daily_units:
    if units < 100:
        continue   # skip the rest of this iteration, go to the next value
    print(f"Reporting: {units} units")
```

### Nested loops

Just like conditionals, loops can be nested — common when you have two dimensions of data, like several stores across several days.


```python
stores = ["Utica", "Albany", "Rome"]
days = ["Mon", "Tue", "Wed"]

for store in stores:
    for day in days:
        print(f"{store} - {day}")
```

### `enumerate()` — getting both the position and the value

Sometimes you need to know *where* you are in a loop, not just the current value. `enumerate()` gives you both at once.


```python
daily_units = [138, 95, 210, 60, 175]

for index, units in enumerate(daily_units):
    print(f"Day {index + 1}: {units} units")   # +1 so we count from Day 1, not Day 0
```

### Exercise 3 — Loop through a week of sales
Using the list below, write a `for` loop that:
1. Prints each day's revenue
2. Keeps a running total
3. After the loop, prints the total and the average


```python
# Exercise 3 
week_revenue = [4521.75, 3980.10, 5102.00, 2870.50, 6023.25, 7110.90, 3340.00]
total_revenue = 0

for day_index, revenue in enumerate(week_revenue, start=1):
    print(f"Day {day_index} Revenue: ${revenue:.2f}")
    total_revenue += revenue

average_revenue = total_revenue / len(week_revenue)

print("-" * 30)
print(f"Total Revenue: ${total_revenue:.2f}")
print(f"Average Daily Revenue: ${average_revenue:.2f}")
```

### Exercise 4 (stretch) — Count days above average
Using the same `week_revenue` list: first compute the average, then loop through again and count how many days were *above* that average. Print the count.


```python
# Exercise 4 
week_revenue = [4521.75, 3980.10, 5102.00, 2870.50, 6023.25, 7110.90, 3340.00]

# 1. Compute the average
avg_revenue = sum(week_revenue) / len(week_revenue)

# 2. Count days above the average
days_above_avg = 0
for revenue in week_revenue:
    if revenue > avg_revenue:
        days_above_avg += 1

print(f"Average revenue: ${avg_revenue:.2f}")
print(f"Days above average: {days_above_avg}")
```

### Exercise 5 (stretch) — Categorize every day at once
Using `week_revenue` again: loop through the list and, for each day, print one of `"Strong day"` (> 5000), `"Average day"` (3000–5000), or `"Slow day"` (< 3000). Then print how many days fell into each category.

*Hint: you'll want three counters, one per category, initialized to 0 before the loop.*


```python
# Exercise 5 
week_revenue = [4521.75, 3980.10, 5102.00, 2870.50, 6023.25, 7110.90, 3340.00]

# Counter initialization
strong_days = 0
average_days = 0
slow_days = 0

for day_index, revenue in enumerate(week_revenue, start=1):
    if revenue > 5000:
        category = "Strong day"
        strong_days += 1
    elif 3000 <= revenue <= 5000:
        category = "Average day"
        average_days += 1
    else:
        category = "Slow day"
        slow_days += 1
        
    print(f"Day {day_index} (${revenue:.2f}): {category}")

print("-" * 30)
print("Summary Count:")
print(f"Strong days (> $5000): {strong_days}")
print(f"Average days ($3000–$5000): {average_days}")
print(f"Slow days (< $3000): {slow_days}")
```
