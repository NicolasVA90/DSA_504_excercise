## Part 4 — Automation with Loops & Conditionals (25%)


**4a.** Write a script called `store_report.sh` that loops over all four stores (`Utica`, `Albany`, `Rome`, `Syracuse`), computes each store's total revenue, and writes one line per store to a file called `store_report.txt` (not the screen — use `>>`).


```python
%%bash
# TODO 4a: Create store_report.sh script to loop through stores, compute revenue, and write to file
cd hw5_work/..
cat > store_report.sh << 'EOF'
#!/bin/bash
# Loop through all four stores and write total revenue to store_report.txt

rm -f store_report.txt
for store in Utica Albany Rome Syracuse; do
    awk -F',' -v s="$store" '$2 == s {sum += $4} END {print s ": " sum}' sales_log.txt >> store_report.txt
done
EOF

chmod +x store_report.sh
./store_report.sh
cat store_report.txt
```

**4b.** Modify your script so that, for any store whose total revenue exceeds **$5000**, the line written to the report includes the label `[HIGH PERFORMER]`. Run the updated script and print the final `store_report.txt` to confirm — you should see at least one store with the label and at least one without it.


```python
%%bash
# TODO 4b: Update store_report.sh to add [HIGH PERFORMER] label if total revenue exceeds $5000
cd hw5_work/..
cat > store_report.sh << 'EOF'
#!/bin/bash
# Loop through stores, calculate revenue, and add [HIGH PERFORMER] label if > 5000

rm -f store_report.txt
for store in Utica Albany Rome Syracuse; do
    awk -F',' -v s="$store" '
    $2 == s {sum += $4} 
    END {
        if (sum > 5000)
            print s ": " sum " [HIGH PERFORMER]";
        else
            print s ": " sum;
    }' sales_log.txt >> store_report.txt
done
EOF

chmod +x store_report.sh
./store_report.sh
cat store_report.txt
```
