## Part 3 — Sort (20%)


**3a.** Sort the full dataset by `revenue` in descending order, and print just the `date`, `store`, `category`, and `revenue` columns for the top 5 rows.


```python
# TODO 3a
# 3a: Sort dataset by revenue descending and display specific columns for top 5 rows
top5_revenue = sales.sort_values(by='revenue', ascending=False)[['date', 'store', 'category', 'revenue']].head(5)
top5_revenue
```

```python
# TODO 3b
# 3b: Sort by store (ascending) and units_sold (descending), showing top 10 rows
sorted_multi = sales.sort_values(by=['store', 'units_sold'], ascending=[True, False])[['store', 'units_sold']].head(10)
sorted_multi
```

## Part 4 — Group & Aggregate (30%)


**4a.** Compute total revenue per store. Which store has the highest total?


```python
# TODO 4a
# 4a: Compute total revenue per store and identify the store with the highest revenue
store_revenue = sales.groupby('store')['revenue'].sum()
top_store = store_revenue.idxmax()

print("Total Revenue per Store:\n", store_revenue)
print("\nHighest Revenue Store:", top_store)
```

**4b.** Compute the average `units_sold` per category (using the raw, uncleaned category column — yes, this means you'll get more than 5 groups, and that's expected for this assignment).


```python
# TODO 4b
# 4b: Compute average units sold per uncleaned category group
avg_units_per_category = sales.groupby('category')['units_sold'].mean()
avg_units_per_category
```

**4c.** Using `.agg()`, compute both the **sum** and the **count** of `revenue` per store in a single call. Then sort the result so the store with the highest total revenue appears first.


```python
# TODO 4c
# 4c: Compute sum and count of revenue per store using .agg(), then sort by total revenue descending
store_revenue_stats = sales.groupby('store')['revenue'].agg(['sum', 'count']).sort_values(by='sum', ascending=False)
store_revenue_stats
```

**4d.** Group by **both** `store` and `category` together, and compute total revenue for each store/category combination. Print just the first 10 rows of the result.


```python
# TODO 4d
# 4d: Group by two columns together ('store' and 'category') and calculate total revenue, showing top 10 rows
store_category_revenue = sales.groupby(['store', 'category'])['revenue'].sum()
store_category_revenue.head(10)
```
