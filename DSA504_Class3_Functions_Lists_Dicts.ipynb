## 1. Functions

A **function** is a named, reusable block of code. Instead of copy-pasting the same logic every time you need it, you write it once and *call* it whenever you need it.

Why this matters: once your scripts get longer than a few lines (which they will, starting today), functions are what keep code organized, testable, and easy to reuse — instead of one long wall of code.


````python
# The simplest possible function -- no inputs, no return value
def greet():
    print("Welcome to the DSA 504 sales report")

# Defining a function doesn't run it -- you have to CALL it
greet()
greet()   # you can call it as many times as you want
````

### Functions with parameters

A **parameter** is an input the function expects. This is what makes functions actually useful — the same function can process different data each time it's called.


````python
def greet_store(store_name):
    print(f"Welcome to the sales report for {store_name}")

greet_store("Utica Main St")
greet_store("Albany Downtown")   # same function, different input
````

### Functions with return values

So far our functions just `print()` something. Usually you want a function to *compute* something and hand the result back to you, so you can use it elsewhere in your code. That's what `return` does.


````python
def calculate_total(units, price_per_unit):
    total = units * price_per_unit
    return total

# The returned value can be stored in a variable
day_total = calculate_total(138, 32.75)
print(day_total)

# Or used directly
print(f"Today's total: ${calculate_total(95, 32.75):.2f}")
````

**Common beginner mistake:** confusing `print()` with `return`. `print()` just displays something on screen — it doesn't give the value back to your program. `return` actually hands the value back so you can use it in further calculations. A function can `print()`, `return`, both, or neither.


````python
# Demonstrating the difference between print and return
def version_print(units, price):
    print(units * price)   # only displays -- can't be reused

def version_return(units, price):
    return units * price   # hands the value back -- CAN be reused

result_a = version_print(100, 10)   # this prints 1000, but result_a is None
result_b = version_return(100, 10)  # this doesn't print anything by itself

print("result_a is:", result_a)   # None -- print() didn't give anything back
print("result_b is:", result_b)   # 1000 -- return DID give something back
````

### Default parameter values

You can give a parameter a default value, which is used automatically if the caller doesn't provide one. This is extremely common for optional settings.


````python
def calculate_total(units, price_per_unit=32.75):
    return units * price_per_unit

# Uses the default price
print(calculate_total(100))

# Overrides the default with a specific price
print(calculate_total(100, 40.00))
````

### Multiple parameters and keyword arguments

Functions can take as many parameters as you need. You can pass values either by position (the order you listed them) or by name (**keyword arguments**), which is often clearer.


````python
def sales_summary(store, units, price_per_unit, is_weekend=False):
    total = units * price_per_unit
    day_type = "weekend" if is_weekend else "weekday"
    return f"{store}: {units} units, ${total:.2f} total ({day_type})"

# Positional arguments -- order matters
print(sales_summary("Utica", 138, 32.75, True))

# Keyword arguments -- order doesn't matter, and it's clearer to read
print(sales_summary(store="Albany", units=95, price_per_unit=28.00, is_weekend=False))
````

### Docstrings

A **docstring** is a short description placed right inside a function, explaining what it does. This becomes important as scripts grow — future you (or a teammate) shouldn't have to read the whole function body just to know what it's for.


````python
def calculate_total(units, price_per_unit):
    """Return the total revenue for a given number of units sold."""
    return units * price_per_unit

# You can view a function's docstring like this:
print(calculate_total.__doc__)
````

## 2. Lists

A **list** is an ordered collection of values. You've already seen lists briefly in Class 2 — today we go deeper into creating, accessing, and modifying them.


````python
# Creating a list
daily_units = [138, 95, 210, 60, 175]
store_names = ["Utica", "Albany", "Rome"]

print(daily_units)
print(store_names)
print(len(daily_units))   # len() tells you how many items are in the list
````

### Indexing — accessing one item by position

Python indexing starts at **0**, not 1. Negative indexes count from the end of the list.


````python
daily_units = [138, 95, 210, 60, 175]

print(daily_units[0])    # 138 -- the FIRST item (index 0)
print(daily_units[2])    # 210 -- the third item
print(daily_units[-1])   # 175 -- the LAST item
print(daily_units[-2])   # 60  -- second-to-last item
````

### Slicing — accessing a range of items

Slicing uses `list[start:stop]` — it includes `start`, but stops *before* `stop`.


````python
daily_units = [138, 95, 210, 60, 175]

print(daily_units[0:2])   # [138, 95] -- items at index 0 and 1 (not 2)
print(daily_units[2:])    # [210, 60, 175] -- from index 2 to the end
print(daily_units[:3])    # [138, 95, 210] -- from the start up to (not including) index 3
print(daily_units[-2:])   # [60, 175] -- the last two items
````

### Modifying a list

Lists are **mutable** — you can change them after creating them, unlike strings or numbers.


````python
daily_units = [138, 95, 210, 60, 175]

# Changing one item by index
daily_units[0] = 150
print(daily_units)

# Adding items
daily_units.append(200)          # adds to the end
print(daily_units)

daily_units.insert(0, 99)        # inserts at a specific position
print(daily_units)

# Removing items
daily_units.remove(200)          # removes the FIRST matching value
print(daily_units)

last_value = daily_units.pop()   # removes and returns the LAST item
print(last_value)
print(daily_units)
````

### Other common list operations


````python
daily_units = [138, 95, 210, 60, 175]

print(sorted(daily_units))          # returns a NEW sorted list (doesn't change the original)
print(daily_units)                  # original is unchanged

daily_units.sort()                  # sorts the list IN PLACE (changes the original)
print(daily_units)

daily_units.sort(reverse=True)      # sort descending
print(daily_units)

print(max(daily_units))             # highest value
print(min(daily_units))             # lowest value
print(sum(daily_units))             # total of all values
print(210 in daily_units)           # checking membership -- True/False
````

## 3. Dictionaries

A **dictionary** stores data as **key-value pairs**, instead of a plain ordered sequence. This is a much more natural way to represent a real-world record — like one row of sales data — because each value has a *label*, not just a position.


````python
# Creating a dictionary
sale = {
    "store": "Utica Main St",
    "date": "2026-09-09",
    "category": "electronics",
    "units": 138,
    "revenue": 4521.75
}

print(sale)
print(type(sale))
````

### Accessing values by key

Instead of a numeric index like a list, you look things up by their key name — which is far more readable.


````python
sale = {
    "store": "Utica Main St",
    "units": 138,
    "revenue": 4521.75
}

print(sale["store"])
print(sale["revenue"])

# .get() is a safer way to access a key -- it won't crash if the key doesn't exist
print(sale.get("category"))            # None, since "category" isn't in this dict
print(sale.get("category", "Unknown")) # "Unknown" -- a fallback default value
````

### Adding, updating, and removing keys


````python
sale = {"store": "Utica Main St", "units": 138}

# Adding a new key
sale["revenue"] = 4521.75
print(sale)

# Updating an existing key
sale["units"] = 150
print(sale)

# Removing a key
del sale["units"]
print(sale)

# Removing a key safely (won't crash if it's already gone)
removed_value = sale.pop("revenue", None)
print(removed_value)
print(sale)
````

### Looping over a dictionary


````python
sale = {
    "store": "Utica Main St",
    "category": "electronics",
    "units": 138,
    "revenue": 4521.75
}

# Looping over keys only
for key in sale:
    print(key)

print("---")

# Looping over keys and values together -- this is the one you'll use most
for key, value in sale.items():
    print(f"{key}: {value}")
````

### Checking whether a key exists


````python
sale = {"store": "Utica Main St", "units": 138}

print("store" in sale)      # True
print("revenue" in sale)    # False
````

## 4. Combining Lists and Dictionaries

Real data is usually a **list of dictionaries** — each dictionary is one record (like a row in a spreadsheet), and the list holds all the records together. This is exactly the shape of data we'll load into pandas starting next class, so getting comfortable with this pattern now will make Class 4 click much faster.


````python
# A list of dictionaries -- this is basically a tiny "table"
sales = [
    {"store": "Utica",  "units": 138, "revenue": 4521.75},
    {"store": "Albany", "units": 95,  "revenue": 3980.10},
    {"store": "Rome",   "units": 210, "revenue": 5102.00},
]

# Looping over the list, and accessing each dictionary's values
for record in sales:
    print(f"{record['store']}: {record['units']} units, ${record['revenue']:.2f}")
````

````python
# Combining everything so far: a function that processes a list of dictionaries
def total_revenue(sales_list):
    """Return the sum of revenue across a list of sale dictionaries."""
    total = 0
    for record in sales_list:
        total += record["revenue"]
    return total

sales = [
    {"store": "Utica",  "units": 138, "revenue": 4521.75},
    {"store": "Albany", "units": 95,  "revenue": 3980.10},
    {"store": "Rome",   "units": 210, "revenue": 5102.00},
]

print(f"Total revenue across all stores: ${total_revenue(sales):.2f}")
````

## 5. Your First Standalone Script — Today's Deliverable

Everything so far today has run inside notebook cells. Real Python work is usually written as a **standalone `.py` script** — a plain text file you run from the terminal, not a notebook. Today's deliverable is writing one of these and committing it to Git.

### What's different about a script vs. a notebook
- A notebook cell runs independently; a script runs **top to bottom, all at once**, exactly like a recipe.
- You write and edit a script in VS Code, not in a notebook.
- You run it from the terminal with `python3 filename.py`, not by clicking "Run."


### Step 1 — Create the file

In VS Code, inside your `dsa504-coursework` repo folder, create a new file called `sales_report.py`.


````python
# Copy everything in THIS cell into sales_report.py in VS Code.
# This combines a function, a list of dictionaries, and a loop --
# everything we covered today.

def total_revenue(sales_list):
    """Return the sum of revenue across a list of sale dictionaries."""
    total = 0
    for record in sales_list:
        total += record["revenue"]
    return total


def top_store(sales_list):
    """Return the store name with the highest revenue."""
    best = sales_list[0]
    for record in sales_list:
        if record["revenue"] > best["revenue"]:
            best = record
    return best["store"]


def main():
    sales = [
        {"store": "Utica",  "units": 138, "revenue": 4521.75},
        {"store": "Albany", "units": 95,  "revenue": 3980.10},
        {"store": "Rome",   "units": 210, "revenue": 5102.00},
    ]

    print("=== DSA 504 Sales Report ===")
    for record in sales:
        print(f"{record['store']}: {record['units']} units, ${record['revenue']:.2f}")

    print(f"\nTotal revenue: ${total_revenue(sales):.2f}")
    print(f"Top store: {top_store(sales)}")


if __name__ == "__main__":
    main()
````

**What's that `if __name__ == "__main__":` line?** It's a standard Python convention meaning "only run `main()` if this file is being run directly, not if it's being imported into another script." You'll see this pattern constantly — for now, just know it belongs at the bottom of a script that has a `main()` function.

### Step 2 — Run it from the terminal


You should see the sales report printed out, ending with the total revenue and top store.
